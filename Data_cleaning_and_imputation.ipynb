##Deriving column

```python
def delivery_cat(minutes):
    if minutes < 60:
        return "Fast"
    elif minutes < 150:
        return "neutral"
    else:
        return "slow"
df["Delivery_performance_category"] = df["Delivery_Time_Min"].apply(delivery_cat)
df["Delivery_performance_category"]
```

```python
def age_cat(age):
    if age <= 30:
        return "Younger_Adult"
    elif age <= 50:
        return "Adult"
    else:
        return "Senior"
df["Age_Group"] = df["Customer_Age"].apply(age_cat)
df["Age_Group"]
```
